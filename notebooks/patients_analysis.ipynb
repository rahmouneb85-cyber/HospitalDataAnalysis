## 1. Load the data

## 3. Visualization : Cholesterol vs Age

```python
import matplotlib.pyplot as plt

df_sorted = df.sort_values("Age")   # tri par âge pour une courbe lisible

plt.plot(df_sorted["Age"], df_sorted["Cholesterol"], marker="o")
plt.xlabel("Age")
plt.ylabel("Cholesterol")
plt.title("Cholesterol vs Age")
plt.show()
```
